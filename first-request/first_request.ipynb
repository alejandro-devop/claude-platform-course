```python
# 1. Cargar la clave de API desde el archivo .env
from dotenv import load_dotenv

load_dotenv()

# load_dotenv() mete las variables del .env en el entorno del proceso.
# La clase Anthropic busca sola la variable ANTHROPIC_API_KEY, por eso
# no hay que pasarle la clave a mano.
```

```python
# 2. Crear el cliente
from anthropic import Anthropic

client = Anthropic()

model = "claude-haiku-4-5"
model
```

```python
# 3. Enviar el mensaje
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is the strongest avenger? Answer in one sentence",
        }
    ],
)

message
```

```python
# 4. Leer solo el texto de la respuesta
message.content[0].text
```
